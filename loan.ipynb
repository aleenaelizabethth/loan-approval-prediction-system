```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=["Loan_Status"])
y = (df["Loan_Status"] == "Y").astype(int)   # 1 = approved, 0 = rejected

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train size:", X_train.shape, " Test size:", X_test.shape)
print("\nTrain class ratio:\n", y_train.value_counts(normalize=True).round(3))
print("\nTest class ratio:\n", y_test.value_counts(normalize=True).round(3))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

num_cols = ["ApplicantIncome", "CoapplicantIncome", "LoanAmount", "Loan_Amount_Term"]
cat_cols = ["Gender", "Married", "Education", "Self_Employed", "Property_Area"]
bin_cols = ["Dependents", "Credit_History"]

preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]), num_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]), cat_cols),
    ("ord", SimpleImputer(strategy="most_frequent"), bin_cols),
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_validate

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, class_weight="balanced"),
    "Decision Tree": DecisionTreeClassifier(max_depth=4, class_weight="balanced", random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42),
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rows = []
for name, model in models.items():
    pipe = Pipeline([("prep", preprocessor), ("model", model)])
    scores = cross_validate(pipe, X_train, y_train, cv=cv,
                            scoring=["accuracy", "f1", "roc_auc"])
    rows.append({
        "Model": name,
        "Accuracy": scores["test_accuracy"].mean(),
        "F1": scores["test_f1"].mean(),
        "ROC-AUC": scores["test_roc_auc"].mean(),
    })

results = pd.DataFrame(rows).round(3)
results
```

```python
from sklearn.metrics import classification_report, roc_auc_score

best = Pipeline([("prep", preprocessor), ("model", models["Logistic Regression"])])
best.fit(X_train, y_train)

pred = best.predict(X_test)
print(classification_report(y_test, pred, target_names=["Rejected", "Approved"]))
print("Test ROC-AUC:", round(roc_auc_score(y_test, best.predict_proba(X_test)[:, 1]), 3))
```

```python
import sys
sys.path.append("..")
from src.features import add_features
from sklearn.preprocessing import FunctionTransformer

feature_step = FunctionTransformer(add_features)
```

```python
num_cols_fe = ["ApplicantIncome", "CoapplicantIncome", "LoanAmount", "Loan_Amount_Term",
               "TotalIncome", "EMI", "Balance_Income", "LoanToIncome",
               "Log_TotalIncome", "Log_LoanAmount"]
bin_cols_fe = ["Dependents", "Credit_History", "Has_Coapplicant"]

preprocessor_fe = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]), num_cols_fe),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore")),
    ]), cat_cols),
    ("ord", SimpleImputer(strategy="most_frequent"), bin_cols_fe),
])
```

```python
from sklearn.model_selection import RepeatedStratifiedKFold

cv_rep = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=42)

rows = []
for name, model in models.items():
    for label, pipe in {
        "Original features": Pipeline([("prep", preprocessor), ("model", model)]),
        "With engineered features": Pipeline([("fe", feature_step), ("prep", preprocessor_fe), ("model", model)]),
    }.items():
        s = cross_validate(pipe, X_train, y_train, cv=cv_rep,
                           scoring=["accuracy", "f1", "roc_auc"])
        rows.append({
            "Model": name,
            "Features": label,
            "Accuracy": s["test_accuracy"].mean(),
            "F1": s["test_f1"].mean(),
            "ROC-AUC": s["test_roc_auc"].mean(),
        })

comparison_fe = pd.DataFrame(rows).round(3)
comparison_fe
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

boost_models = {
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=150, max_depth=2, learning_rate=0.05, random_state=42),
    "XGBoost": XGBClassifier(n_estimators=200, max_depth=2, learning_rate=0.05, subsample=0.8,
                             colsample_bytree=0.8, eval_metric="logloss", random_state=42),
    "LightGBM": LGBMClassifier(n_estimators=200, max_depth=3, num_leaves=7, learning_rate=0.05,
                               min_child_samples=20, subsample=0.8, colsample_bytree=0.8,
                               random_state=42, verbose=-1),
}

rows = []
for name, model in boost_models.items():
    pipe = Pipeline([("fe", feature_step), ("prep", preprocessor_fe), ("model", model)])
    s = cross_validate(pipe, X_train, y_train, cv=cv_rep, scoring=["accuracy", "f1", "roc_auc"])
    rows.append({"Model": name,
                 "Accuracy": s["test_accuracy"].mean(),
                 "F1": s["test_f1"].mean(),
                 "ROC-AUC": s["test_roc_auc"].mean()})

boost_results = pd.DataFrame(rows).round(3)
boost_results
```

```python
from sklearn.model_selection import RandomizedSearchCV

search_space = {
    "Random Forest": (
        RandomForestClassifier(class_weight="balanced", random_state=42),
        {
            "model__n_estimators": [200, 300, 500],
            "model__max_depth": [3, 4, 5, 6, None],
            "model__min_samples_leaf": [1, 3, 5, 10],
            "model__max_features": ["sqrt", "log2", 0.5],
        },
    ),
    "XGBoost": (
        XGBClassifier(eval_metric="logloss", random_state=42),
        {
            "model__n_estimators": [100, 200, 300],
            "model__max_depth": [2, 3, 4],
            "model__learning_rate": [0.01, 0.03, 0.05, 0.1],
            "model__subsample": [0.6, 0.8, 1.0],
            "model__colsample_bytree": [0.6, 0.8, 1.0],
            "model__min_child_weight": [1, 3, 5],
        },
    ),
}

tuned = {}
for name, (model, params) in search_space.items():
    pipe = Pipeline([("fe", feature_step), ("prep", preprocessor_fe), ("model", model)])
    search = RandomizedSearchCV(pipe, params, n_iter=30, scoring="roc_auc",
                                cv=cv, random_state=42, n_jobs=-1)
    search.fit(X_train, y_train)
    tuned[name] = search
    print(f"{name}: best CV ROC-AUC = {search.best_score_:.3f}")
    print("  ", search.best_params_)
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import f1_score

best_name = max(tuned, key=lambda k: tuned[k].best_score_)
best_model = tuned[best_name].best_estimator_
print("Winner:", best_name)

oof_proba = cross_val_predict(best_model, X_train, y_train, cv=cv, method="predict_proba")[:, 1]

thresholds = np.arange(0.30, 0.81, 0.01)
macro_f1 = [f1_score(y_train, (oof_proba >= t).astype(int), average="macro") for t in thresholds]
best_threshold = thresholds[int(np.argmax(macro_f1))]
print(f"Chosen threshold: {best_threshold:.2f}  (macro F1 = {max(macro_f1):.3f})")

plt.plot(thresholds, macro_f1)
plt.axvline(best_threshold, color="red", linestyle="--")
plt.xlabel("Threshold"); plt.ylabel("Macro F1 (out-of-fold)")
plt.title("Threshold Selection")
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

test_proba = best_model.predict_proba(X_test)[:, 1]
test_pred = (test_proba >= best_threshold).astype(int)

print(classification_report(y_test, test_pred, target_names=["Rejected", "Approved"]))
print("Test ROC-AUC:", round(roc_auc_score(y_test, test_proba), 3))

ConfusionMatrixDisplay(confusion_matrix(y_test, test_pred),
                       display_labels=["Rejected", "Approved"]).plot(cmap="Blues")
plt.title(f"{best_name}: Test Confusion Matrix")
plt.show()
```

```python
import shap

def get_shap_values(model, X_raw):
    X_fe = model.named_steps["fe"].transform(X_raw)
    X_t = model.named_steps["prep"].transform(X_fe)
    if hasattr(X_t, "toarray"):
        X_t = X_t.toarray()
    names = [n.split("__", 1)[-1] for n in model.named_steps["prep"].get_feature_names_out()]
    explainer = shap.TreeExplainer(model.named_steps["model"])
    sv = explainer.shap_values(X_t)
    ev = explainer.expected_value
    if isinstance(sv, list):          # older shap, Random Forest
        sv, ev = sv[1], ev[1]
    elif sv.ndim == 3:                # newer shap, Random Forest
        sv, ev = sv[:, :, 1], ev[1]
    return sv, float(np.ravel(ev)[0]), X_t, names

sv, base_value, X_test_t, feature_names = get_shap_values(best_model, X_test)

shap.summary_plot(sv, X_test_t, feature_names=feature_names, show=False)
plt.title("What drives approval decisions (SHAP)")
plt.show()

shap.summary_plot(sv, X_test_t, feature_names=feature_names, plot_type="bar")
```
